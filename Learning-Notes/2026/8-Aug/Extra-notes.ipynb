#### Slicing 
----------------------------------
##### When we want only a slice of the string
##### It is defined with colon ":"
##### 1. si[:] --> Print all value
##### 2. si[1:4] --> Print value from 1st to 4th index

##### Slicing has 2 parts:
###### 1. Forward Slicing (Left to Right) (Right most index is called as n-1)
        - Starts with 0

###### 2. Backward Slicing (Right to Left) (Left most index is called as n+1)
        - Starts with -1
