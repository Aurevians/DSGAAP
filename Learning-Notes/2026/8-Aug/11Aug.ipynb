### Slicing
==========

```python
#####      0  1  2  3  4  5  6
##### si = w  e  l  c  o  m  e
#####     -7 -6 -5 -4 -3 -2 -1
```

```python
si = 'welcome'
```

#### Forward Slicing

```python
si[0:7]
```

```python
si[:6]
```

#### Backward Slicing

```python
si[:-1]
```

```python
si[-5:-1]
```

#### Step Slicing or Advanced Slicing:
======================================
##### 1. Forward Step Slicing

```python
si[::2]
```

##### 2. Backward Step Slicing or print the string in reverse order

```python
si[::-1]
```

```python
#####      0  1  2  3  4  5  6
##### si = w  e  l  c  o  m  e
#####     -7 -6 -5 -4 -3 -2 -1
```

```python
si[::-2]
```

#### 1. Arithmatic Operators --> +, -, / , //, %, **
=========================================================

```python
x1=10
x2=20
x3=2
x4=3
print(f'x1+x2 is {x1+x2}')
print(f'x1-x2 is {x1-x2}')
print(f'x1/x2 is {x1/x2}')
print(f'x1//x2 is {x1//x2}')
print(f'x1%x2 is {x1%x2}')
print(f'x1**x2 is {x1**x2}') #10 to the power of 20
print(f'x3**x4 is {x3**x4}') #2 to the power of 3 = 8
print(f'x1*x2 is {x1*x2}')
```
