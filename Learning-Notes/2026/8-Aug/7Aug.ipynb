Python Index begins with 0 (Default)
It has 2 parts
1. Forward Index (Left to Right)
- Starts with 0

2. Backward Index (Right to Left)
- Starts with -1

3. Step Index



### String Indexes

```python
si = "welcome"
# w   e   l   c   o   m   e
# 0   1   2   3   4   5   6
print (si[0])
print (si[1])
print (si[6])
print (si[7])
```

```python
print (si[-1])
print (si[-5])
print (si[-8])
```

### Slicing 

When we want only a slice of the string
It is defined with colon ":"
1. si[:] --> Print all value
2. si[1:4] --> Print value from 1st to 4th index

Slicing has 2 parts:
1. Forward Slicing (Left to Right)
- Starts with 0

2. Backward Slicing (Right to Left)
- Starts with -1


```python
print (si[:])
```

```python
print (si[1:4])
```

```python
si[3:7]
```

```python
si[3:10] # It gives no error, as it is just printing the slice of the string
```

```python
si[10] # It gives error, as there is no index for 10
```

```python
si[3:5]
```

```python
si[2:] # Print the slice from 2nd index to end as Right index is not defined
```
