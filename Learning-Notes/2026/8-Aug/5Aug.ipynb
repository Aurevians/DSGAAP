### NOTES

#### * Rules for Python Variables
======================================
##### * variables are case sesitive
##### * variables cannot start with number
##### * variables cannot use special characters except '_'
##### * variables cannot be declared using keyword or reserve words

#### * Types of Errors
======================================
##### * SyntaxError (When Sytax is not right)
##### * NameError (When variable is not defined)
