### Slicing
================

```python
si = 'welcome'
```

```python
si
```

```python
#####      0  1  2  3  4  5  6
##### si = w  e  l  c  o  m  e
#####     -7 -6 -5 -4 -3 -2 -1
```

#### Forward Slicing:
----------------------

```python
si[0:7]
```

```python
si[:7]
```

```python
# if nothing is mentioned for left/right it is considered as zero or the last index
```

```python
si[0:6]
```

```python
# Slice from 0 to 5 (n-1) index
```

```python
si[0:]
```

```python
si[5:]
```

#### Backward Slicing:
----------------------

```python
si[:-1]
```

```python
# Slice from -7 to -2 (-n-1) index
```

```python
si[-8:]
```

```python
# Slice from -7 to -1 index (till end)
```

#### Step Slicing or Advanced Slicing:
----------------------------------------
##### 1. Forward Step Slicing

```python
si = 'welcome'
```

```python
#####      0  1  2  3  4  5  6
##### si = w  e  l  c  o  m  e
#####     -7 -6 -5 -4 -3 -2 -1
```

```python
si[0:7:1]
```

```python
si[0:6:1]
```

```python
si [::2]
```

##### 2. Backward Step Slicing or print the string in reverse order

```python
si[::-1]
```

```python
si[::-2]
```
